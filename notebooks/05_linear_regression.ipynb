```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
data = pd.read_csv(
    "../datasets/processed/ml_dataset.csv"
)

print("Dataset shape:", data.shape)
```

```python
X = data[
    [
        "dividend_yield",
        "dividend_growth",
        "dividend_consistency",
        "return_volatility"
    ]
]

y = data["total_return"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

model = LinearRegression()

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("Model recreated successfully!")
```

```python
mae = mean_absolute_error(y_test, y_pred)

mse = mean_squared_error(y_test, y_pred)

rmse = np.sqrt(mse)

r2 = r2_score(y_test, y_pred)

print("Linear Regression Evaluation")
print("-----------------------------")
print(f"MAE  : {mae:.4f}")
print(f"MSE  : {mse:.4f}")
print(f"RMSE : {rmse:.4f}")
print(f"R²   : {r2:.4f}")
```

## Evaluation Metrics Interpretation

### MAE
Mean Absolute Error measures the average absolute difference between actual and predicted total returns.

MAE = 12.8967 means that the predictions differ from the actual returns by approximately 12.9 percentage points on average.

### MSE
Mean Squared Error gives more importance to large prediction errors.

MSE = 303.6466.

### RMSE
Root Mean Squared Error is easier to interpret because it is in the same unit as the target variable.

RMSE = 17.4255 means the typical prediction error is approximately 17.43 percentage points.

### R² Score
R² measures how much variation in total return is explained by the model.

R² = 0.3665, meaning the model explains approximately 36.65% of the variation in total return on this test set.

```python
comparison = pd.DataFrame({
    "Actual_Return": y_test.values,
    "Predicted_Return": y_pred,
    "Residual": y_test.values - y_pred
})

print(comparison.head(10).to_string(index=False))
```

```python
print("Residual Statistics")
print("-------------------")
print(comparison["Residual"].describe())
```

```python
print("Mean residual:", comparison["Residual"].mean())
```
