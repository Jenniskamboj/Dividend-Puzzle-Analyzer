```python
def get_fundamentals(ticker):
    try:
        stock = yf.Ticker(ticker)

        financials = stock.financials
        balance_sheet = stock.balance_sheet

        if financials.empty:
            return pd.DataFrame()

        # Revenue
        if "Total Revenue" in financials.index:
            revenue = financials.loc["Total Revenue"].copy()
        else:
            revenue = pd.Series(dtype=float)

        # Net Income
        if "Net Income" in financials.index:
            net_income = financials.loc["Net Income"].copy()
        else:
            net_income = pd.Series(dtype=float)

        # Assets
        if "Total Assets" in balance_sheet.index:
            total_assets = balance_sheet.loc["Total Assets"].copy()
        else:
            total_assets = pd.Series(dtype=float)

        # Convert dates to years
        revenue.index = pd.to_datetime(
            revenue.index
        ).year

        net_income.index = pd.to_datetime(
            net_income.index
        ).year

        total_assets.index = pd.to_datetime(
            total_assets.index
        ).year

        result = pd.DataFrame({
            "year": revenue.index,
            "revenue": revenue.values
        })

        result["net_income"] = (
            result["year"]
            .map(net_income)
        )

        result["total_assets"] = (
            result["year"]
            .map(total_assets)
        )

        result["ticker"] = ticker

        return result[
            [
                "ticker",
                "year",
                "revenue",
                "net_income",
                "total_assets"
            ]
        ]

    except Exception as e:
        print(f"Error for {ticker}: {e}")
        return pd.DataFrame()
```

```python
all_fundamentals = []

for ticker in companies["ticker"].dropna():
    print("Collecting:", ticker)

    result = get_fundamentals(ticker)

    if not result.empty:
        all_fundamentals.append(result)
```

```python
fundamentals = pd.concat(
    all_fundamentals,
    ignore_index=True
)

fundamentals.head()
```
