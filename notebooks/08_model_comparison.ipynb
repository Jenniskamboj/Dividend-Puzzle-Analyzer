```python
import pandas as pd
import numpy as np
import os

rf_file = "../datasets/processed/day19_random_forest_results.csv"

print("File exists:", os.path.exists(rf_file))

rf_results = pd.read_csv(rf_file)

print(rf_results)
```

```python
# Linear Regression results
linear_results = pd.DataFrame({
    "Model": ["Linear Regression"],
    "MAE": [12.896688748879361],
    "MSE": [303.6465688580839],
    "RMSE": [17.425457493508855],
    "R2": [0.3665095339776454]
})

# Random Forest results
rf_results = pd.read_csv(
    "../datasets/processed/day19_random_forest_results.csv"
)

# XGBoost results
xgb_results = pd.DataFrame({
    "Model": ["XGBoost"],
    "MAE": [15.1574],
    "MSE": [403.3640],
    "RMSE": [20.0839],
    "R2": [0.1585]
})

# Combine all models
comparison = pd.concat(
    [
        linear_results,
        rf_results[["Model", "MAE", "MSE", "RMSE", "R2"]],
        xgb_results
    ],
    ignore_index=True
)

print(comparison)
```

```python
comparison.round(4)
```

```python
best_mae = comparison.loc[
    comparison["MAE"].idxmin(), "Model"
]

best_rmse = comparison.loc[
    comparison["RMSE"].idxmin(), "Model"
]

best_r2 = comparison.loc[
    comparison["R2"].idxmax(), "Model"
]

print("Best model based on MAE :", best_mae)
print("Best model based on RMSE:", best_rmse)
print("Best model based on R²  :", best_r2)
```

```python
comparison.to_csv(
    "../datasets/processed/day21_model_comparison.csv",
    index=False
)

print("Day 21 comparison saved successfully!")
```

## Final Model Selection

Based on the evaluation results, Linear Regression was selected as the best-performing model.

### Final Results

| Model | MAE | RMSE | R² |
|---|---:|---:|---:|
| Linear Regression | 12.8967 | 17.4255 | 0.3665 |

Linear Regression achieved the lowest MAE and RMSE and the highest R² score.

Therefore, Linear Regression will be used as the primary predictive model for the subsequent Dividend Puzzle theory analysis.
