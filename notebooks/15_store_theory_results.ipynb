```python
import pandas as pd
import os
```

```python
theory_file = "../datasets/processed/day27_theory_comparison.csv"

print("File exists:", os.path.exists(theory_file))

theory_results = pd.read_csv(theory_file)

print(theory_results)
```

```python
best_theory = theory_results.loc[
    theory_results["Rank"].idxmin()
]

print("Best Theory:", best_theory["Theory"])
print("Evidence Score:", best_theory["Evidence_Score"])
print("Rank:", best_theory["Rank"])
```

```python
week4_summary = pd.DataFrame({
    "Metric": [
        "Best Performing Theory",
        "Best Theory Score"
    ],
    "Value": [
        best_theory["Theory"],
        best_theory["Evidence_Score"]
    ]
})

print(week4_summary)
```

```python
week4_summary.to_csv(
    "../datasets/processed/week4_final_summary.csv",
    index=False
)

print("Week 4 summary saved successfully!")
```
